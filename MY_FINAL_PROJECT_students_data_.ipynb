DATA SET CLEANING

```python
missing_values = df.isnull().sum()
print("missing values:\n",missing_values)
df.fillna(df.mean(), inplace=True)
```

REMOVING DUPLICATES


```python
df.duplicated()
```

```python
df = df.drop_duplicates()
```

```python
df.shape
```

DATA MANIPULATING


```python
df['Total']=df['Maths']+df['English']+df['Science']+df['computer']+df['History']
df.head()
```

```python
df['percentage']=(df['Total']/500)*100
df.head()
```

```python
df = df.drop(['percentage'], axis=1)
df.head(4)
```

DATA ANALYSIS


```python
weak_performers = df.nsmallest(5, 'Total')
print("weak Performers:\n", weak_performers[['ID', 'Total']])
```

```python
top_performers = df.nlargest(5, 'Total')
print("Top performers:\n" top_performers[['ID', 'Total']])
```

```python
import matplotlib.pyplot as plt
```

```python
average_success_rate_math = df.groupby('Maths')['y'].mean().mean()
average_success_rate_english = df.groupby('English')['y'].mean().mean()
average_success_rate_science = df.groupby('Science')['y'].mean().mean()
average_success_rate_computer = df.groupby('computer')['y'].mean().mean()
average_success_rate_history = df.groupby('History')['y'].mean().mean()
```

```python
subjects = ['Math', 'English', 'Science', 'Computer', 'History']
average_success_rates = [average_success_rate_math,
average_success_rate_english,
average_success_rate_science,
average_success_rate_computer,
average_success_rate_history]
```

```python
subjects = ['Math', 'English', 'Science', 'Computer', 'History']
average_success_rates = [average_success_rate_math,
average_success_rate_english,
average_success_rate_science,
average_success_rate_computer,
average_success_rate_history]
```

```python
plt.figure(figsize=(8, 8))
plt.pie(average_success_rates, labels=subjects, autopct='%1.1f%%', colors=['skyblue', 'lightcoral', 'mediumseagreen', 'gold', 'lightcoral'])
plt.title('Average Success Rate by Subject')
plt.show()
```

DATA VISUALIZATION

```python
success_rate_gender = df.groupby('gender')['y'].mean()
success_rate_race = df.groupby('race')['y'].mean()
# Plot success rates by gender
plt.figure(figsize=(10, 5))
success_rate_gender.plot(kind='bar', color='skyblue')
plt.title('Success Rate by Gender')
plt.xlabel('Gender')
plt.ylabel('Success Rate')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
df[df['y'] == 1]['Maths'].plot(kind='hist', alpha=0.5, color='red', label='Failure', bins=20)
df[df['y'] == 0]['Maths'].plot(kind='hist', alpha=0.5, color='skyblue', label='Success', bins=20)
plt.title("Distribution of Math Curriculum Scores for Success and Failure")
plt.xlabel('Math Curriculum Scores')
plt.ylabel('Number of Students')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(10, 5))
df.groupby(['class', 'y']).size().unstack().plot(kind='bar', stacked=True, color=['skyblue', 'lightcoral'])
plt.title("Influence of Student Class on Graduate School Applications")
plt.xlabel('Class')
plt.ylabel('Number of Students')
plt.show()
```

```python
df.groupby(['from4', 'y']).size().unstack().plot(kind='bar', stacked=True, color=['skyblue', 'lightcoral'])
plt.title("Influence of Family Background on Graduate School Applications")
plt.xlabel('Family Background Indicator')
plt.ylabel('Number of Students')
plt.show()
```

```python
df.groupby(['from1', 'y']).size().unstack().plot(kind='bar', stacked=True, color=['skyblue', 'lightcoral'])
plt.title("Impact of Home Locations on Graduate School Applications")
plt.xlabel('Home Location')
plt.ylabel('Number of Students')
plt.show()

```
